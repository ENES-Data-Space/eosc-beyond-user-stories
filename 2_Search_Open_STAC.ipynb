# Search & Access climate data
This notebook queries the ENES catalog (https://catalogue.eneslab.pilot.eosc-beyond.eu/) and retrieves the Zarr asset of the selected resource.

## Data Search

```python
import requests
import pandas as pd
from pystac_client import Client
import json

url_stac = "https://api.eneslab.pilot.eosc-beyond.eu"

variable = "tas"
experiment = "historical"
frequency = "day"
```

```python
catalog = Client.open(url_stac)

search = catalog.search(
    collections=["CMIP_S3"],
    query={
        "cmip6:experiment_id": {"eq": experiment},
        "cmip6:variable_id": {"eq": variable},
        "cmip6:table_id": {"eq": frequency},
    },
    method="GET"
)
```

## Data Selection

```python
items = list(search.items())
items
```

```python
item = items[0]
zarr_url = item.assets["zarr"].href
zarr_url
```

# Accessing provenance documents

## 1. Through the yProv API

### Get the list of available documents

```python
import requests, json
yprov_endpoint = "http://yprov.disi.unitn.it:8004/"
url = yprov_endpoint + "documents"
headers = {
    "Content-Type": "application/json",
}

response = requests.get(url, headers=headers,timeout=10)
for item in response.json():
    print(item)
```

### Retrieve a document and store it to a JSON file

```python
pid = "..."
url = yprov_endpoint + "documents/"+pid
headers = {
    "Content-Type": "application/json",
}

response = requests.get(url, headers=headers)
data = response.json()

with open('./exported.json', 'w', encoding='utf-8') as file:
    json.dump(data, file, ensure_ascii=False, indent=4)
```

### Display the provenance document

```python
from prov.model import ProvDocument
from prov.dot import prov_to_dot

with open("./exported.json", "r", encoding="utf-8") as json_file:
    json_data = json_file.read()
    
d2 = ProvDocument.deserialize(content=json_data, format="json")
dot = prov_to_dot(d2)
dot.write_png('./exported.png')
```
